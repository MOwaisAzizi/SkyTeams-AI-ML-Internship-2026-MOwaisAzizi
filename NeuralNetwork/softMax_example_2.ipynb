```python
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense
from sklearn.datasets import make_blobs


# 1. Create 4-class dataset
centers = [[-5, 2], [-2, -2], [1, 2], [5, -2]]

X_train, y_train = make_blobs(
    n_samples=200,
    centers=centers,
    cluster_std=1.0,
    random_state=30
)


# 2. Plot original data
plt.scatter(X_train[:, 0], X_train[:, 1], c=y_train)
plt.xlabel("x0")
plt.ylabel("x1")
plt.title("Original Data")
plt.show()


# 3. Create model
model = Sequential([
    Dense(2, activation="relu"),
    Dense(4, activation="linear")
])


# 4. Compile and train
model.compile(
    loss=tf.keras.losses.SparseCategoricalCrossentropy(from_logits=True),
    optimizer=tf.keras.optimizers.Adam(0.01)
)

model.fit(X_train, y_train, epochs=200, verbose=0)


# 5. Get Layer 1 output
W1, b1 = model.layers[0].get_weights()

X_new = np.maximum(0, X_train @ W1 + b1)


# 6. Plot Layer 1 output
plt.scatter(X_new[:, 0], X_new[:, 1], c=y_train)
plt.xlabel("a₀[1]")
plt.ylabel("a₁[1]")
plt.title("New Features → Input to Layer 2")
plt.show()


# 7. See some examples
print("Original X:")
print(X_train[:5])

print("\nLayer 1 output:")
print(X_new[:5])


# 8. Final predictions
logits = model.predict(X_train[:5], verbose=0)

print("\nLayer 2 output:")
print(logits)

print("\nPredicted classes:")
print(np.argmax(logits, axis=1))
```
