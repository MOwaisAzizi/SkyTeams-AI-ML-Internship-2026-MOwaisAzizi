```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from sklearn.datasets import make_blobs

# ---- 1. create test dataset around this poins(four class) ----
centers = [[-5, 2], [-2, -2], [1, 2], [5, -2]]
X_train, y_train = make_blobs(n_samples=2000, centers=centers,
                               cluster_std=1.0, random_state=30)

# ---- 2. Model with SOFTMAX output (outputs are already probabilities) ----
model = Sequential([
    Dense(25, activation='relu'),
    Dense(15, activation='relu'),
    Dense(4,  activation='softmax')     # <-- softmax applied here
])

# ---- 3. Loss reads probabilities directly (no extra softmax needed) ----
model.compile(
    loss=tf.keras.losses.SparseCategoricalCrossentropy(),  # from_logits defaults to False
    optimizer=tf.keras.optimizers.Adam(0.001),
)

model.fit(X_train, y_train, epochs=10, verbose=0)

# ---- 4. Predictions are ALREADY probabilities ----
probs = model.predict(X_train[:2])
print("probabilities:\n", probs)
print("sum of each row:", np.sum(probs, axis=1))   # sums to 1, no extra step needed

# ---- 5. Predicted class — same as before ----
predicted_class = np.argmax(probs, axis=1)
print("predicted classes:", predicted_class)
```
