```python
import numpy as np

def sigmoid(z):
    return 1 / (1 + np.exp(-z))

g = sigmoid

# -----------------------------------------
# Vectorized dense layer function
# -----------------------------------------
def dense(A_in, W, b):
    Z = np.matmul(A_in, W) + b
    A_out = g(Z)
    return A_out


# -----------------------------------------
# Weights and biases (same as coffee lab, W1/b1)
# -----------------------------------------
W1 = np.array([[-8.93,  0.29, 12.9],
               [-0.1,  -7.32, 10.81]])   # shape (2, 3)
b1 = np.array([-9.82, -9.28, 0.96])       # shape (3,)

W2 = np.array([[-31.18], [-27.59], [-32.56]])   # shape (3, 1)
b2 = np.array([15.41])                           # shape (1,)


# -----------------------------------------
# Input: single example, double brackets = shape (1, 2)
# -----------------------------------------
A_in = np.array([[200, 17]])

# -----------------------------------------
# Full forward pass — no loop anywhere
# -----------------------------------------
A1 = dense(A_in, W1, b1)   # layer 1 output, shape (1, 3)
A2 = dense(A1, W2, b2)     # layer 2 output, shape (1, 1)

print("Layer 1 output:", A1)
print("Final prediction:", A2)
```
