```python
from pinecone import Pinecone, ServerlessSpec

# 1. Initialize Pinecone
pc = Pinecone(api_key="YOUR_API_KEY")

# 2. Example vector data with metadata
vectors = [
    {
        "id": "vec1",
        "values": [0.1] * 1536,
        "metadata": {"text": "I love machine learning"}
    },
    {
        "id": "vec2",
        "values": [0.2] * 1536,
        "metadata": {"text": "Python is great for AI"}
    },
    {
        "id": "vec3",
        "values": [0.3] * 1536,
        "metadata": {"text": "Pinecone stores embeddings"}
    }
]

# 3. Check that ALL vectors have 1536 dimensions
vector_dims = all(
    [len(vector["values"]) == 1536 for vector in vectors]
)

print("All vectors valid:", vector_dims)

# 4. Create a Pinecone index
pc.create_index(
    name="datacamp-index",
    dimension=1536,
    spec=ServerlessSpec(
        cloud="aws",
        region="us-east-1"
    )
)

# 5. Connect to the existing index
index = pc.Index("datacamp-index")

# 6. Upload vectors and metadata
index.upsert(vectors=vectors)

# 7. Print index statistics
print(index.describe_index_stats())
```
